## Base NLP Implementation

```python
import numpy as np
import math
from collections import Counter
```

```python
documents = [
    "i love nlp nlp",
    "you love ml"
]

tokenized = [doc.split() for doc in documents]

vocab = sorted(set(word for doc in tokenized for word in doc))
vocab_index = {w: i for i, w in enumerate(vocab)}

def tf(doc):
    counts = Counter(doc)
    total = len(doc)
    return {w: counts[w] / total for w in counts}

def idf(tokenized_docs):
    N = len(tokenized_docs)
    df = Counter(word for doc in tokenized_docs for word in set(doc))
    return {w: math.log(N / df[w]) for w in df}

idf_vals = idf(tokenized)

tfidf = []
for doc in tokenized:
    tf_vals = tf(doc)
    vec = [tf_vals.get(w, 0) * idf_vals.get(w, 0) for w in vocab]
    tfidf.append(vec)

cos_sim = dict()
for i in range(len(tokenized)):
    for j in range(i, len(tokenized)):
        if tokenized[i] != tokenized[j]:
            counts1 = Counter(tokenized[i])
            counts2 = Counter(tokenized[j])
            cos_sim[(i+1, j+1)] = np.dot(counts1, counts2)/(np.linalg.norm(counts1)*np.linalg.norm(counts2))


            

print("Vocabulary:", vocab)
print("TF-IDF vectors:", tfidf)
print("Cosine Similarity between docs:", cos_sim)
```
